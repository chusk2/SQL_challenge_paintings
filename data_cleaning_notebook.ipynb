## `product_size` dataset

```python
product_size = pd.read_csv('product_size.csv')
```

```python
product_size.head()
```

### missing values

```python
product_size.isna().sum()
```

### duplicates

```python
product_size.duplicated().sum()
```

```python
product_size[product_size.duplicated(keep = False)].sort_values(['work_id', 'size_id'])
```

### drop duplicates

```python
product_size = product_size.drop_duplicates()
```

### dtypes

```python
product_size.dtypes
```

`size_id` appears with object dtype. It must be due to missing values for `size_id`

```python
product_size.size_id.unique()
```

There are some value that don't make sense, since they have decimal places. Other values are `#VALUE!`. Let's check what percentage the wrong values make out of the total dataset.

```python
(round( product_size[product_size.size_id.str.contains('.', regex = False)]
       .shape[0] / product_size.shape[0] * 100, 3) )
```

```python
round( product_size[product_size.size_id == '#VALUE!']
      .shape[0] / product_size.shape[0] * 100, 3)
```

```python
anomalous_size_id_records = product_size[(product_size.size_id == "#VALUE!") | 
                                         (product_size.size_id.str.contains('.', regex = False))]

round( anomalous_size_id_records.shape[0] / product_size.shape[0] * 100, 3)
```

Since the records in `product_size` dataset which have anomalous values for `size_id` (`#VALUE!` or decimal values) don't make up more than 0.3% of the total dataset, the easiest and quickest way of dealing with them is to remove them. The time spent trying to figure out how to clean this values is not worth.

```python
product_size.loc[anomalous_size_id_records.index, 'size_id'] = np.nan
```

```python
product_size.isna().sum()
```

After cleaning and removing the anomalous values, I will convert the `size_id` column to `float` dtype.

```python
product_size.loc[:, 'size_id'] = product_size.size_id.astype('float')
```

Now, I want to check if there are any records whose `size_id` value don't have a match in the `size_id` column from the `canvas_size` dataset.

```python
canvas = pd.read_csv('canvas_size.csv')
```

I will left merge the `product_size` dataset with the `canvas_size`, using `size_id` as key column. If there were any record with a `size_id` which doesn't exist in the `canvas_size` dataset, their `width` values would be `NaN`.

```python
prod_canvas_merged = product_size.dropna(subset='size_id').merge(canvas, how = 'left', on = 'size_id')
```

```python
prod_canvas_merged.width.isna().sum()
```

Finally, I can say that the cleaning process of `product_size` dataset is finished.

### Export `product_size` dataset

```python
product_size.to_csv('../clean_csv_datasets/product_size.csv', index = False, na_rep='NULL')
```

### conversion of opening and closing hours from **AM / PM** format into **24H** format.

For the times using *PM*, I will add 12 hours to the hour number.

```python
def clean_hours(time_str):
    hour = time_str[:2]
    minutes = time_str[3:5]
    day_period = time_str[-2:]

    if day_period == 'PM':
        clean_time = f'{int(hour) + 12:02}:{minutes}'
    elif day_period == 'AM':
        clean_time = hour + ':' + minutes

    return clean_time
```

```python
museum_hours.open.sample(5)
```

```python
museum_hours.close.sample(5)
```

Clean the hours

```python
museum_hours.loc[:, 'open'] = museum_hours.open.apply(clean_hours)
museum_hours.loc[:, 'close'] = museum_hours.close.apply(clean_hours)
```
